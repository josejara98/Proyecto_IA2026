## Datos del Dataset


```python

from pathlib import Path
from collections import Counter
import yaml
import pandas as pd

ruta = Path("/content/frutilla_v3-8")

# Leer las clases
with open(ruta / "data.yaml", "r") as f:
    config = yaml.safe_load(f)

clases = config["names"]
print("CLASES:", clases)

resultados = []

for conjunto in ["train", "valid", "test"]:
    carpeta_img = ruta / conjunto / "images"
    carpeta_lbl = ruta / conjunto / "labels"

    imagenes = [
        p for p in carpeta_img.iterdir()
        if p.suffix.lower() in [".jpg", ".jpeg", ".png"]
    ]

    conteo = Counter()

    for archivo in carpeta_lbl.glob("*.txt"):
        for linea in archivo.read_text().splitlines():
            if linea.strip():
                clase = int(linea.split()[0])
                conteo[clase] += 1

    fila = {
        "Conjunto": conjunto,
        "Imágenes": len(imagenes)
    }

    for i, nombre in enumerate(clases):
        fila[nombre] = conteo[i]

    resultados.append(fila)

tabla = pd.DataFrame(resultados)
display(tabla)
```

### Visualización de 6 imágnes etiquetadas con sus cuadros delimitadores

```python

import random
import yaml
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from PIL import Image
from pathlib import Path

ruta = Path("/content/frutilla_v3-8")

with open(ruta / "data.yaml", "r") as f:
    config = yaml.safe_load(f)

clases = config["names"]

if isinstance(clases, dict):
    clases = {int(k): v for k, v in clases.items()}
else:
    clases = dict(enumerate(clases))

colores = {
    "maduro": "red",
    "no-maduro": "lime",
    "semi-maduro": "orange"
}

carpeta = ruta / "train" / "images"

imagenes = [
    p for p in carpeta.iterdir()
    if p.suffix.lower() in [".jpg", ".jpeg", ".png"]
]

random.seed(42)
seleccionadas = random.sample(
    imagenes, min(6, len(imagenes))
)

fig, axes = plt.subplots(2, 3, figsize=(15, 10))

for ax, imagen_path in zip(axes.flat, seleccionadas):

    imagen = Image.open(imagen_path).convert("RGB")
    ancho, alto = imagen.size

    ax.imshow(imagen)

    etiqueta = (
        ruta / "train" / "labels" /
        (imagen_path.stem + ".txt")
    )

    if etiqueta.exists():
        for linea in etiqueta.read_text().splitlines():

            datos = linea.split()

            if len(datos) != 5:
                continue

            clase = int(datos[0])
            x, y, w, h = map(float, datos[1:])

            x1 = (x - w/2) * ancho
            y1 = (y - h/2) * alto

            nombre = clases[clase]
            color = colores.get(nombre, "cyan")

            rect = patches.Rectangle(
                (x1, y1),
                w * ancho,
                h * alto,
                linewidth=1.5,
                edgecolor=color,
                facecolor="none"
            )

            ax.add_patch(rect)
            ax.text(
                x1, max(0, y1 - 3),
                nombre,
                color="white",
                fontsize=8,
                bbox=dict(facecolor=color, alpha=0.8)
            )

    ax.set_title(imagen_path.name, fontsize=9)
    ax.axis("off")

plt.tight_layout()
plt.show()
```

## Inferencia del modelo sobre un video entre los segundos 20 y 40

```python

import cv2
from pathlib import Path

# Configuración del recorte
inicio_segundos = 20
fin_segundos = 40

# Ruta del video original
# Ya definida anteriormente como ruta_video

# Archivo de salida
ruta_salida = "/content/frutillas_inferencia_20_40.mp4"

cap = cv2.VideoCapture(ruta_video)

if not cap.isOpened():
    raise RuntimeError("No se pudo abrir el video")

fps = cap.get(cv2.CAP_PROP_FPS)
total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

if fps <= 0:
    raise RuntimeError("FPS del video no válido")

duracion = total_frames / fps

print(f"Duración original: {duracion:.2f} segundos")

if inicio_segundos < 0 or fin_segundos <= inicio_segundos:
    raise ValueError("Intervalo de tiempo inválido")

if fin_segundos > duracion:
    raise ValueError("El video no llega al segundo 40")

ancho = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
alto = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

# Convertir segundos a fotogramas
frame_inicio = int(inicio_segundos * fps)
frame_fin = int(fin_segundos * fps)

# Ir directamente al segundo 20
cap.set(cv2.CAP_PROP_POS_FRAMES, frame_inicio)

# Crear video de salida
fourcc = cv2.VideoWriter_fourcc(*"mp4v")

out = cv2.VideoWriter(
    ruta_salida,
    fourcc,
    fps,
    (ancho, alto)
)

if not out.isOpened():
    cap.release()
    raise RuntimeError("No se pudo crear el video de salida")

frame_actual = frame_inicio

while frame_actual < frame_fin:

    ret, frame = cap.read()

    if not ret:
        break

    # Inferencia con YOLO11n
    resultados = modelo.predict(
        source=frame,
        imgsz=512,
        conf=0.25,
        iou=0.7,
        device=0,
        verbose=False
    )

    # Dibujar cajas de detección
    frame_anotado = resultados[0].plot()

    # Guardar fotograma
    out.write(frame_anotado)

    frame_actual += 1

    if (frame_actual - frame_inicio) % 100 == 0:
        print(
            f"Procesados: "
            f"{frame_actual-frame_inicio}/"
            f"{frame_fin-frame_inicio}"
        )

cap.release()
out.release()

print("Inferencia finalizada")
print("Video generado:", ruta_salida)
```

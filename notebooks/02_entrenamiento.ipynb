```python
import os
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score, f1_score
from ultralytics import YOLO

# Rutas configurables
BASE_DIR = Path("..")
DATA_DIR = BASE_DIR / "data"
MODELS_DIR = BASE_DIR / "models"
RESULTS_DIR = BASE_DIR / "results"

FEATURES_TRAIN = DATA_DIR / "features_train.csv"
FEATURES_VAL = DATA_DIR / "features_val.csv"
METRICAS_CSV = RESULTS_DIR / "metricas.csv"

CLASES = ["maduro", "no-maduro", "semi-maduro"]
print("Entorno configurado. Directorio de resultados:", RESULTS_DIR)
```

## 1. Clasificador Base Auxiliar: Regresión Logística sobre Características HSV

Entrenamos la Regresión Logística base con regularización =1.0$ y pesos de clase uniformes (class_weight=None) sobre las características normalizadas.


```python
if FEATURES_TRAIN.exists() and FEATURES_VAL.exists():
    df_train = pd.read_csv(FEATURES_TRAIN)
    df_val = pd.read_csv(FEATURES_VAL)
    
    features = ["red_ratio", "green_ratio", "mean_saturation"]
    
    X_train = df_train[features].values
    y_train = df_train["class_id"].values
    
    X_val = df_val[features].values
    y_val = df_val["class_id"].values
    
    print(f"Muestras de entrenamiento: {len(X_train)} | Muestras de validación: {len(X_val)}")
    
    # Entrenar modelo base
    modelo_base = LogisticRegression(C=1.0, class_weight=None, max_iter=1000, random_state=42)
    modelo_base.fit(X_train, y_train)
    
    # Evaluación sobre validación
    y_pred = modelo_base.predict(X_val)
    
    print("\n=== REPORTE DE CLASIFICACIÓN — REGRESIÓN LOGÍSTICA BASE ===")
    print(classification_report(y_val, y_pred, target_names=CLASES, digits=4))
    print(f"Exactitud global (Accuracy): {accuracy_score(y_val, y_pred):.4f}")
    print(f"F1 Macro: {f1_score(y_val, y_pred, average='macro'):.4f}")
else:
    print("Archivos de características no encontrados en data/. Verifique la ejecución de 01_preprocesamiento.")
```

## 2. Reconstrucción de Código de Entrenamiento Reproducible de Detectores YOLO

A continuación se presenta la formulación de entrenamiento empleada en Google Colab con aceleración GPU Tesla T4.
Ambos modelos usan el optimizador **AdamW**, tamaño de entrada 640, batch 16, tasa de aprendizaje inicial 0.001 y parada temprana con paciencia de 20 épocas.


```python
# Código de entrenamiento para Modelo 1 (YOLO11s — frutilla_v3)
codigo_entrenamiento_m1 = '''
from ultralytics import YOLO

model = YOLO("yolo11s.pt")
results = model.train(
    data="data/modelo1/data.yaml",
    epochs=100,
    imgsz=640,
    batch=16,
    device=0,
    optimizer="AdamW",
    lr0=0.001,
    weight_decay=0.0005,
    patience=20,
    seed=42,
    deterministic=True,
    plots=True,
    save=True,
    project="experimentos_frutillas",
    name="E01_YOLO11s",
    exist_ok=True
)
'''

# Código de entrenamiento para Modelo 2 (YOLO11n — frutilla_v4)
codigo_entrenamiento_m2 = '''
from ultralytics import YOLO

model = YOLO("yolo11n.pt")
results = model.train(
    data="data/modelo2/data.yaml",
    epochs=100,
    imgsz=640,
    batch=16,
    device=0,
    optimizer="AdamW",
    lr0=0.001,
    weight_decay=0.0005,
    patience=20,
    seed=42,
    deterministic=True,
    plots=True,
    save=True,
    project="experimentos_frutillas",
    name="E01_YOLO11n",
    exist_ok=True
)
'''
print("Código de entrenamiento para ambas variantes disponible para ejecución en GPU.")
```

```python
import os
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import f1_score, accuracy_score, precision_score, recall_score, classification_report

BASE_DIR = Path("..")
DATA_DIR = BASE_DIR / "data"
RESULTS_DIR = BASE_DIR / "results"

FEATURES_TRAIN = DATA_DIR / "features_train.csv"
FEATURES_VAL = DATA_DIR / "features_val.csv"
OPTIMIZACION_CSV = RESULTS_DIR / "optimizacion.csv"

CLASES = ["maduro", "no-maduro", "semi-maduro"]
```

## 1. Ejecución de Grid Search sobre Regresión Logística (HSV)


```python
if FEATURES_TRAIN.exists() and FEATURES_VAL.exists():
    df_train = pd.read_csv(FEATURES_TRAIN)
    df_val = pd.read_csv(FEATURES_VAL)
    
    features = ["red_ratio", "green_ratio", "mean_saturation"]
    X_train, y_train = df_train[features].values, df_train["class_id"].values
    X_val, y_val = df_val[features].values, df_val["class_id"].values
    
    c_candidates = [0.01, 0.1, 1.0, 10.0]
    cw_candidates = [None, "balanced"]
    
    resultados = []
    
    for c in c_candidates:
        for cw in cw_candidates:
            clf = LogisticRegression(C=c, class_weight=cw, max_iter=1000, random_state=42)
            clf.fit(X_train, y_train)
            y_pred = clf.predict(X_val)
            
            f1_clases = f1_score(y_val, y_pred, average=None)
            res = {
                "modelo": "RegresionLogistica_HSV",
                "C": c,
                "class_weight": str(cw),
                "val_accuracy": round(accuracy_score(y_val, y_pred), 4),
                "val_precision_macro": round(precision_score(y_val, y_pred, average='macro', zero_division=0), 4),
                "val_recall_macro": round(recall_score(y_val, y_pred, average='macro', zero_division=0), 4),
                "val_f1_macro": round(f1_score(y_val, y_pred, average='macro'), 4),
                "f1_maduro": round(f1_clases[0], 4),
                "f1_no_maduro": round(f1_clases[1], 4),
                "f1_semi_maduro": round(f1_clases[2], 4)
            }
            resultados.append(res)
            
    df_opt = pd.DataFrame(resultados)
    df_opt.to_csv(OPTIMIZACION_CSV, index=False)
    print("Búsqueda Grid Search finalizada y guardada en:", OPTIMIZACION_CSV)
    display(df_opt.sort_values(by="val_f1_macro", ascending=False))
```

## 2. Comparación entre Modelo Base y Modelo Optimizado

Identificamos la mejor combinación según F1 macro en validación y la comparamos con el modelo inicial sin balanceo.


```python
if OPTIMIZACION_CSV.exists():
    df_opt = pd.read_csv(OPTIMIZACION_CSV)
    mejor = df_opt.loc[df_opt["val_f1_macro"].idxmax()]
    base = df_opt[(df_opt["C"] == 1.0) & (df_opt["class_weight"] == "None")].iloc[0]
    
    print("=== COMPARATIVA: MODELO BASE vs MODELO OPTIMIZADO ===")
    print(f"Modelo Base      (C=1.0, class_weight=None):     F1 Macro = {base['val_f1_macro']:.4f} | F1 Semi-maduro = {base['f1_semi_maduro']:.4f}")
    print(f"Modelo Optimizado(C={mejor['C']}, class_weight={mejor['class_weight']}): F1 Macro = {mejor['val_f1_macro']:.4f} | F1 Semi-maduro = {mejor['f1_semi_maduro']:.4f}")
    print(f"\nIncremento relativo de F1 Macro: +{((mejor['val_f1_macro'] - base['val_f1_macro']) / base['val_f1_macro']) * 100:.2f}%")
```

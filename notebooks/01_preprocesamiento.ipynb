## 1. Verificación de Integridad de Datasets y Detección de Incidencias

Comprobamos que:
- Cada imagen tenga su archivo de anotación homónimo .txt.
- No existan coordenadas fuera de $[0.0, 1.0]$ o dimensiones no positivas.
- No existan clases fuera de $\{0, 1, 2\}$.


```python
def auditar_dataset(split_dir):
    img_dir = Path(split_dir) / "images"
    lbl_dir = Path(split_dir) / "labels"
    
    if not img_dir.exists() or not lbl_dir.exists():
        print(f"Ruta no encontrada: {split_dir}")
        return
        
    imagenes = list(img_dir.glob("*.jpg")) + list(img_dir.glob("*.png"))
    etiquetas = list(lbl_dir.glob("*.txt"))
    
    incidencias = {"faltan_etiquetas": 0, "cajas_invalidas": 0, "clases_desconocidas": 0}
    conteo_clases = Counter()
    total_cajas = 0
    
    for img_path in imagenes:
        lbl_path = lbl_dir / f"{img_path.stem}.txt"
        if not lbl_path.exists():
            incidencias["faltan_etiquetas"] += 1
            continue
            
        with open(lbl_path, "r") as f:
            for linea in f:
                partes = linea.strip().split()
                if len(partes) != 5:
                    incidencias["cajas_invalidas"] += 1
                    continue
                c, x, y, w, h = int(partes[0]), float(partes[1]), float(partes[2]), float(partes[3]), float(partes[4])
                if c not in CLASES:
                    incidencias["clases_desconocidas"] += 1
                else:
                    conteo_clases[c] += 1
                if not (0 <= x <= 1 and 0 <= y <= 1 and 0 < w <= 1 and 0 < h <= 1):
                    incidencias["cajas_invalidas"] += 1
                total_cajas += 1
                
    print(f"=== Auditoría [{Path(split_dir).name}] ===")
    print(f"Imágenes: {len(imagenes)} | Anotaciones: {len(etiquetas)} | Total instancias: {total_cajas}")
    print(f"Incidencias: {incidencias}")
    for c_id, nom in CLASES.items():
        print(f" - {nom} (clase {c_id}): {conteo_clases[c_id]}")

# Ejecutar sobre la partición de prueba local incluida en el repositorio
auditar_dataset(DATASET_M1 / "test")
```

## 2. Definición y Fundamentación de las 3 Características HSV

Para satisfacer el requerimiento de ingeniería de características a partir de los datos crudos, se formulan tres métricas calculadas sobre el espacio de color HSV (Hue, Saturation, Value):

1. **ed_ratio (Proporción de píxeles rojos)**:
   - El rojo en HSV se localiza en los extremos del círculo cromático:  \in [0^\circ, 20^\circ] \cup [330^\circ, 360^\circ]$ (en OpenCV con rango  \in [0, 179]$, corresponde a $[0, 10] \cup [165, 179]$).
   - Se exige un umbral mínimo de saturación ( \ge 0.25$) y brillo ( \ge 0.20$) para excluir reflejos especulares y zonas en sombra extrema.
   - **Utilidad**: Es la característica más correlacionada con el estado de madurez comercial óptimo (maduro).

2. **green_ratio (Proporción de píxeles verdes)**:
   - El verde se ubica en  \in [70^\circ, 165^\circ]$ (OpenCV: $[35, 85]$).
   - Umbrales:  \ge 0.20$,  \ge 0.20$.
   - **Utilidad**: Cuantifica la presencia de clorofila activa presente en frutos inmaduros (
o-maduro) o en la corona/hombros de frutos en viraje (semi-maduro).

3. **mean_saturation (Saturación media normalizada)**:
   - Media de la componente  \in [0, 1]$ en toda el área del recorte.
   - **Utilidad**: A medida que la frutilla madura, los pigmentos antociánicos se concentran incrementando notablemente la pureza o vivacidad del color respecto a las etapas previas.


```python
def extraer_caracteristicas_hsv(crop_bgr):
    """
    Calcula las tres características HSV a partir del recorte BGR de un fruto.
    """
    if crop_bgr is None or crop_bgr.size == 0:
        return 0.0, 0.0, 0.0
        
    hsv = cv2.cvtColor(crop_bgr, cv2.COLOR_BGR2HSV)
    H = hsv[:, :, 0]
    S = hsv[:, :, 1] / 255.0
    V = hsv[:, :, 2] / 255.0
    
    total_pix = crop_bgr.shape[0] * crop_bgr.shape[1]
    if total_pix == 0:
        return 0.0, 0.0, 0.0
        
    # Máscara de rojo
    mascara_rojo = ((H <= 10) | (H >= 165)) & (S >= 0.25) & (V >= 0.20)
    red_ratio = np.sum(mascara_rojo) / total_pix
    
    # Máscara de verde
    mascara_verde = (H >= 35) & (H <= 85) & (S >= 0.20) & (V >= 0.20)
    green_ratio = np.sum(mascara_verde) / total_pix
    
    # Saturación media
    mean_sat = float(np.mean(S))
    
    return round(float(red_ratio), 4), round(float(green_ratio), 4), round(float(mean_sat), 4)

print("Función de extracción HSV definida correctamente con umbrales conservados.")
```

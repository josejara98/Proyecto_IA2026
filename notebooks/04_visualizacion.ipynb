## Objetivos del Notebook

2. **Curvas Reales de Convergencia por Época**: Graficar las curvas de pérdida (ox_loss, cls_loss, dfl_loss) y métricas de validación (mAP50, mAP50-95, Box(P), Box(R)) a partir de los registros de 38 y 61 épocas.

5. **Comparativa Crítica Modelo 1 vs Modelo 2**: Contrastar el aporte de los datos externos frente a la arquitectura, advirtiendo sobre las diferencias en las particiones de validación.


```python
import os
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from ultralytics import YOLO

BASE_DIR = Path("..")
DATA_DIR = BASE_DIR / "data"
MODELS_DIR = BASE_DIR / "models"
RESULTS_DIR = BASE_DIR / "results"
PLOTS_DIR = RESULTS_DIR / "plots"

LOG_M1 = RESULTS_DIR / "training_log_yolo11s.csv"
LOG_M2 = RESULTS_DIR / "training_log_yolo11n.csv"
```

## 1. Curvas de Convergencia de Entrenamiento por Época (Registros Históricos Reales)

Trazamos la evolución de mAP50, mAP50-95 y las pérdidas de entrenamiento de ambas variantes.


```python
if LOG_M1.exists() and LOG_M2.exists():
    df_m1 = pd.read_csv(LOG_M1)
    df_m2 = pd.read_csv(LOG_M2)
    
    fig, axes = plt.subplots(2, 2, figsize=(15, 10))
    
    # Pérdida de caja
    axes[0, 0].plot(df_m1["epoch"], df_m1["box_loss"], label="Modelo 1 (YOLO11s - Propios)", color="blue")
    axes[0, 0].plot(df_m2["epoch"], df_m2["box_loss"], label="Modelo 2 (YOLO11n - Externos)", color="orange")
    axes[0, 0].set_title("Evolución de Pérdida de Caja (Box Loss)")
    axes[0, 0].set_xlabel("Época")
    axes[0, 0].set_ylabel("Pérdida")
    axes[0, 0].grid(True, linestyle="--", alpha=0.5)
    axes[0, 0].legend()
    
    # Pérdida de clasificación
    axes[0, 1].plot(df_m1["epoch"], df_m1["cls_loss"], label="Modelo 1 (YOLO11s)", color="blue")
    axes[0, 1].plot(df_m2["epoch"], df_m2["cls_loss"], label="Modelo 2 (YOLO11n)", color="orange")
    axes[0, 1].set_title("Evolución de Pérdida de Clase (Cls Loss)")
    axes[0, 1].set_xlabel("Época")
    axes[0, 1].grid(True, linestyle="--", alpha=0.5)
    axes[0, 1].legend()
    
    # mAP50
    axes[1, 0].plot(df_m1["epoch"], df_m1["mAP50"], label="Modelo 1 (YOLO11s)", color="blue")
    axes[1, 0].plot(df_m2["epoch"], df_m2["mAP50"], label="Modelo 2 (YOLO11n)", color="orange")
    axes[1, 0].set_title("Evolución de mAP @ IoU=0.50")
    axes[1, 0].set_xlabel("Época")
    axes[1, 0].set_ylabel("mAP50")
    axes[1, 0].grid(True, linestyle="--", alpha=0.5)
    axes[1, 0].legend()
    
    # mAP50-95
    axes[1, 1].plot(df_m1["epoch"], df_m1["mAP50_95"], label="Modelo 1 (YOLO11s)", color="blue")
    axes[1, 1].plot(df_m2["epoch"], df_m2["mAP50_95"], label="Modelo 2 (YOLO11n)", color="orange")
    axes[1, 1].set_title("Evolución de mAP @ IoU=[0.50:0.95]")
    axes[1, 1].set_xlabel("Época")
    axes[1, 1].set_ylabel("mAP50-95")
    axes[1, 1].grid(True, linestyle="--", alpha=0.5)
    axes[1, 1].legend()
    
    plt.tight_layout()
    plt.show()
```

## 2. Matrices de Confusión de Detección

Visualizamos las matrices de confusión normalizadas generadas por Ultralytics durante las corridas de validación.
El umbral de IoU de evaluación es **0.50** y la fila/columna ackground cuantifica las detecciones espurias (falsos positivos) y los frutos no detectados (falsos negativos).


```python
cm_m1 = PLOTS_DIR / "modelo1" / "confusion_matrix_normalized.png"
cm_m2 = PLOTS_DIR / "modelo2" / "confusion_matrix_normalized.png"

fig, axes = plt.subplots(1, 2, figsize=(16, 7))

if cm_m1.exists():
    axes[0].imshow(Image.open(cm_m1))
    axes[0].set_title("Matriz de Confusión Normalizada — Modelo 1 (YOLO11s)")
    axes[0].axis("off")

if cm_m2.exists():
    axes[1].imshow(Image.open(cm_m2))
    axes[1].set_title("Matriz de Confusión Normalizada — Modelo 2 (YOLO11n)")
    axes[1].axis("off")

plt.tight_layout()
plt.show()
```

## 3. Inferencia de Detección y Visualización de Cuadros Delimitadores

Ejecutamos inferencia sobre imágenes de prueba conservadas en data/modelo1/test/images/.


```python
peso_m1 = MODELS_DIR / "best_YOLO11s.pt"
peso_m2 = MODELS_DIR / "best_YOLO11n.pt"
test_dir = DATA_DIR / "modelo1" / "test" / "images"

if peso_m1.exists() and test_dir.exists():
    imgs_prueba = list(test_dir.glob("*.jpg"))[:3]
    modelo = YOLO(str(peso_m1))
    
    for img_p in imgs_prueba:
        res = modelo.predict(source=str(img_p), conf=0.25, imgsz=640, verbose=False)[0]
        im_anotada = res.plot()
        
        plt.figure(figsize=(8, 8))
        plt.imshow(im_anotada[..., ::-1])
        plt.title(f"Predicciones YOLO11s sobre {img_p.name}")
        plt.axis("off")
        plt.show()
```

## 4. Histograma del Error de Conteo por Imagen

Calculamos el error de conteo ( = \text{conteo predicho} - \text{conteo real anotado}$) para cada imagen de prueba.


```python
lbl_dir = DATA_DIR / "modelo1" / "test" / "labels"

if peso_m1.exists() and test_dir.exists() and lbl_dir.exists():
    modelo = YOLO(str(peso_m1))
    errores_conteo = []
    
    for img_p in test_dir.glob("*.jpg"):
        lbl_p = lbl_dir / f"{img_p.stem}.txt"
        n_real = 0
        if lbl_p.exists():
            with open(lbl_p, "r") as f:
                n_real = sum(1 for line in f if line.strip())
                
        res = modelo.predict(source=str(img_p), conf=0.25, imgsz=640, verbose=False)[0]
        n_pred = len(res.boxes)
        errores_conteo.append(n_pred - n_real)
        
    plt.figure(figsize=(9, 4))
    plt.hist(errores_conteo, bins=np.arange(min(errores_conteo)-0.5, max(errores_conteo)+1.5, 1), 
             color="teal", edgecolor="black", alpha=0.7)
    plt.axvline(0, color="red", linestyle="--", label="Error cero (Conteo exacto)")
    plt.title("Histograma del Error de Conteo por Imagen (Test Set)")
    plt.xlabel("Error de Conteo (Predicho - Real)")
    plt.ylabel("Cantidad de Imágenes")
    plt.legend()
    plt.grid(True, linestyle="--", alpha=0.5)
    plt.show()
```

## 5. Comparativa Crítica entre Variantes de Modelo

### Conclusiones Metodológicas:
1. **Efecto de la Diversidad de Datos**: A pesar de utilizar una arquitectura Nano con solo el 27% de los parámetros y el 30% del costo computacional de YOLO11s, el Modelo 2 superó en todas las métricas al Modelo 1 gracias a la mayor riqueza de fondos, iluminaciones y variaciones morfológicas provistas por los datos externos.
2. **Advertencia sobre Validación Cruzada**: Las particiones de validación son distintas (215 vs 1,042 imágenes). Para aislar con certeza si la mejora proviene exclusivamente de los datos externos o del balance relativo de clases, se recomienda evaluar ambos detectores sobre el conjunto de prueba propio (	est de rutilla_v3).
